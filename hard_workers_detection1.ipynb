```python
# Import required libraries
from ultralytics import YOLO

# Load a YOLOv8 model (Pretrained)
model = YOLO('yolov8s.pt')  # Use 'yolov8n.pt' for a smaller model

# Train the model
model.train(data='/content/Hard-Hat-Workers-1/data.yaml', epochs=50, batch=16, imgsz=640)

# Validate the model
metrics = model.val()

# Run inference on an image
results = model('image.jpg')

# Show results
results.show()
```

```python

import cv2
import torch
import glob
import os
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load the YOLOv8 model
model = YOLO('yolov8s.pt')  # You can use 'yolov8n.pt' for a smaller version

# Path to the folder containing test images
test_folder = "/content/Hard-Hat-Workers-1/test/images"  # Change to your folder path

# Get all image files in the folder
image_paths = glob.glob(os.path.join(test_folder, "*.jpg"))  # Adjust the extension if needed

# Process each image
for image_path in image_paths:
    results = model(image_path)  # Run YOLOv8 inference

    # Display results for each image
    for result in results:
        img = result.plot()  # Draw bounding boxes

        # Show the image
        plt.figure(figsize=(10, 10))
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        plt.axis("off")
        plt.title(f"Detection: {os.path.basename(image_path)}")
        plt.show()
```
